#Fórmula de las Diferencias Divididas de Newton
### Algoritmo 3.2 (Burden & Faires, p. 93) — Ejemplo 1 (p. 94)

---

## **1. Formulación del Problema y Método**

Dados $n + 1$ puntos distintos $(x_0, f(x_0)), \dots, (x_n, f(x_n))$, el polinomio interpolante $P_n(x)$ en la **forma progresiva de Newton** es:

$$P_n(x) = F_{0,0} + \sum_{i=1}^{n} F_{i,i} \prod_{j=0}^{i-1} (x - x_j)$$

El **Algoritmo 3.2** calcula los coeficientes $F_{i,i}$ construyendo una tabla triangular inferior mediante:

* **Orden $0$:** $F_{i,0} = f(x_i)$ para $i = 0, 1, \dots, n$.
* **Orden $j$ ($j \ge 1$):**
  $$F_{i,j} = \frac{F_{i,j-1} - F_{i-1,j-1}}{x_i - x_{i-j}}, \quad i = 1, \dots, n; \quad j = 1, \dots, i$$

Los coeficientes buscados corresponden a la **diagonal principal** $(F_{0,0}, F_{1,1}, \dots, F_{n,n})$.

---

## **2. Datos de Prueba: Ejemplo 1 (Burden, p. 94, Tabla 3.10)**

| $i$ | $x_i$ | $f(x_i)$ |
| :---: | :---: | :---: |
| $0$ | $1.0$ | $0.7651977$ |
| $1$ | $1.3$ | $0.6200860$ |
| $2$ | $1.6$ | $0.4554022$ |
| $3$ | $1.9$ | $0.2818186$ |
| $4$ | $2.2$ | $0.1103623$ |


```python
import numpy as np  # 'as' asigna un alias corto al módulo


def algoritmo_diferencias_divididas(x, fx):
    """Calcula la tabla y los coeficientes F_{i,i} de Newton (Algoritmo 3.2)."""
    n = len(x) - 1
    # 'np.zeros((filas, cols))' inicializa una matriz vacía en punto flotante
    F = np.zeros((n + 1, n + 1), dtype=float)

    # Slicing '[:, 0]': asigna fx a todas las filas de la columna 0
    F[:, 0] = fx

    # 'range(1, n + 1)' genera enteros de 1 hasta n (límite superior exclusivo)
    for i in range(1, n + 1):
        for j in range(1, i + 1):
            F[i, j] = (F[i, j - 1] - F[i - 1, j - 1]) / (x[i] - x[i - j])

    # 'np.diag(F)' extrae directamente la diagonal principal [F_{0,0}, ..., F_{n,n}]
    return np.diag(F), F


# Datos del Ejemplo 1 (Burden, p. 94, Tabla 3.10)
x_datos = [1.0, 1.3, 1.6, 1.9, 2.2]
fx_datos = [0.7651977, 0.6200860, 0.4554022, 0.2818186, 0.1103623]

# Desempaquetado de la tupla retornada por la función
coeficientes, tabla_F = algoritmo_diferencias_divididas(x_datos, fx_datos)

print("TABLA 3.11: DIFERENCIAS DIVIDIDAS DE NEWTON")
print("-" * 68)
print(f"{'i':<3} {'x_i':<6} {'F[i,0]':>10} {'F[i,1]':>11} {'F[i,2]':>11} {'F[i,3]':>11} {'F[i,4]':>11}")
print("-" * 68)

for i in range(len(x_datos)):
    # List comprehension con f-strings para formatear cada fila a 7 decimales
    fila = " ".join(f"{tabla_F[i, j]:>11.7f}" for j in range(i + 1))
    print(f"{i:<3} {x_datos[i]:<6.1f} {fila}")

print("\nCOEFICIENTES DEL POLINOMIO (Diagonal F_{i,i}):")
# 'enumerate()' obtiene simultáneamente el índice 'i' y el valor 'c'
for i, c in enumerate(coeficientes):
    print(f"  F[{i},{i}] = {c:.7f}")
```
