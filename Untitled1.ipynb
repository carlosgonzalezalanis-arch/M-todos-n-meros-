# Método de Newton

## 1. Formulación del Problema

De acuerdo con el Ejemplo 1, el problema consiste en aproximar la raíz de la función:
$$ f(x) = \cos(x) - x = 0 $$

Para aplicar el método, calculamos la derivada de la función:
$$ f'(x) = -\sin(x) - 1 $$

## 2. Exposición del Método

El algoritmo iterativo del método de Newton parte de una aproximación inicial $p_0$ y calcula las siguientes aproximaciones mediante la fórmula:
$$ p_n = p_{n-1} - \frac{f(p_{n-1})}{f'(p_{n-1})}, \quad \text{para } n \geq 1 $$

El proceso se detiene cuando se alcanza un número máximo de iteraciones $N_0$ o cuando el error absoluto es menor a una tolerancia $TOL$:
$$ |p_n - p_{n-1}| < TOL $$


```python
import numpy as np

def f(x):
    return np.cos(x) - x

def df(x):
    return -np.sin(x) - 1

def metodo_newton(p0, TOL, N0):
    print(f"{'Iteración':<10} | {'Aproximación p':<18} | {'Error'}")
    print("-" * 50)


    i = 1

    while i <= N0:

        p = p0 - f(p0) / df(p0)
        error = abs(p - p0)
        print(f"{i:<10} | {p:<18.10f} | {error:.5e}")
        if error < TOL:
            print("-" * 50)
            # Solución final resaltada
            print(f"Solución final encontrada: p = {p:.10f}")
            return p
        i += 1
        p0 = p

    #  Mensaje de error si se excede el número máximo de iteraciones
    print(f"El método falló tras {N0} iteraciones.")
    return None

# Definición de parámetros iniciales y llamada a la función
p0 = np.pi / 4
TOL = 1e-5
N0 = 20

raiz = metodo_newton(p0, TOL, N0)
```

```python
import matplotlib.pyplot as plt

x = np.linspace(0, 1.5, 100)
y = f(x)

plt.figure(figsize=(8, 5))

plt.plot(x, y, color='blue', label=r'$f(x) = \cos(x) - x$')

plt.axhline(0, color='black', linestyle='--')


if raiz is not None:
    plt.plot(raiz, 0, marker='o', color='red', markersize=8, label=f'Raíz: {raiz:.6f}')

plt.title('Solución Gráfica del Método de Newton')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()

plt.show()
```
